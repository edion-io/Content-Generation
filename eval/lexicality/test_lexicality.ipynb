## 2. Compare long sentences

```python
from itertools import combinations

for c in combinations(long_sents, 2):
  print("Sentence 1:", c[0], "\n\n")
  print("Sentence 2:", c[1])
  # Test lexicality in both ways
  l1 = lex.lexicality([c[0]], [c[1]])
  l2 = lex.lexicality([c[1]], [c[0]])

  print("FRE Similarity s1 -> s2", l1['fre_similarity'])
  print("Rouge1 s1 -> s2: ", l1['rouge1'])
  print("RougeS s1 -> s2: ", l1['rougeS'])

  print()
```

Overall we can see that FRE similarity works well, for example for the last two sentences, they have an FRE similarity of 0.78 as sentence 1 has 4 more monosyllabic words and 4/20 ~= 0.15, which makes sense as to why the FRE differs by that number then. 

For rouge 1 and rougeS, we want a high Rouge1 and low RougeS or the inverse, as a it would mean that the structure is similar but with different words, or that the model is using a similar vocab but in a different way (and both of these outcomes are viable). We can see for example, that in this case the two sentences have a low score for both as they are quite different. Rouge tends to be rather sensitive so we still find some similarity in those sentences, mainly due to the determiners they use.
